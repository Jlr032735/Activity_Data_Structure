# Exemplos da Aula — Listas Simplesmente Encadeadas

Ordem dos exemplos:

## Exemplo 01 — Montando a primeira cadeia de músicas

**Origem na apresentação:** slide 5.

Cria três nós independentes, conecta-os e percorre a playlist seguindo a referência `proximo` até `None`.


```python
# Exemplo 01 - Montando a primeira cadeia de músicas
# Conceitos: nó, referência, encadeamento e percurso.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


musica1 = No("Imagine")
musica2 = No("Yellow")
musica3 = No("Believer")

musica1.proximo = musica2
musica2.proximo = musica3

inicio = musica1

print("=== PLAYLIST ENCADEADA ===")

atual = inicio

while atual is not None:
    print(f"♪ {atual.musica}")
    atual = atual.proximo

print("FIM")
```

## Exemplo 02 — Criando uma música como nó

**Origem na apresentação:** slide 7.

Mostra que um nó pode existir antes de ser conectado à estrutura: ele armazena o dado e inicialmente possui `proximo = None`.


```python
# Exemplo 02 - Criando uma música como nó
# Conceito: um nó pode existir antes de ser conectado à lista.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


novo_no = No("Imagine")

print("=== NOVO NÓ ===")
print("Música :", novo_no.musica)
print("Próximo:", novo_no.proximo)
```

## Exemplo 03 — Apontando para a próxima faixa

**Origem na apresentação:** slide 9.

Evidencia a alteração de uma referência: Imagine começa sem sucessor e depois passa a apontar para Yellow.


```python
# Exemplo 03 - Apontando para a próxima faixa
# Conceito: uma referência liga um nó ao seu sucessor.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


primeira = No("Imagine")
segunda = No("Yellow")

print("=== ANTES DA LIGAÇÃO ===")
print("Música atual       :", primeira.musica)
print("Próximo de Imagine :", primeira.proximo)

primeira.proximo = segunda

print("\n=== DEPOIS DA LIGAÇÃO ===")
print("Música atual       :", primeira.musica)
print("Próximo de Imagine :", primeira.proximo.musica)
```

## Exemplo 04 — Percorrendo a playlist de conexão em conexão

**Origem na apresentação:** slide 11.

Percorre os nós a partir do início e avança por `atual.proximo` até alcançar o final da lista.


```python
# Exemplo 04 - Percorrendo a playlist de conexão em conexão
# Conceito: percurso sequencial seguindo a referência proximo.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


n1 = No("Imagine")
n2 = No("Yellow")
n3 = No("Believer")

n1.proximo = n2
n2.proximo = n3

inicio = n1
atual = inicio
contador = 1

print("=== PERCURSO DA PLAYLIST ===")

while atual is not None:
    print(f"{contador}ª música: {atual.musica}")
    atual = atual.proximo
    contador += 1

print("Fim da lista: None")
```

## Exemplo 05 — Inserindo Paradise entre duas músicas

**Origem na apresentação:** slide 13.

Insere um novo nó entre Yellow e Believer preservando primeiro a continuação da cadeia e depois alterando a referência de Yellow.


```python
# Exemplo 05 - Inserindo Paradise entre duas músicas
# Conceito: inserir um nó alterando referências, sem deslocar os demais nós.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


def exibir(inicio):
    atual = inicio

    while atual is not None:
        print(atual.musica, end="")
        if atual.proximo is not None:
            print(" -> ", end="")
        atual = atual.proximo

    print(" -> None")


imagine = No("Imagine")
yellow = No("Yellow")
believer = No("Believer")

imagine.proximo = yellow
yellow.proximo = believer

print("=== ANTES DA INSERÇÃO ===")
exibir(imagine)

novo = No("Paradise")

# Primeiro preservamos a continuação da lista.
novo.proximo = yellow.proximo

# Depois Yellow passa a apontar para Paradise.
yellow.proximo = novo

print("\n=== DEPOIS DA INSERÇÃO ===")
exibir(imagine)
```

## Exemplo 06 — Removendo Paradise sem quebrar a playlist

**Origem na apresentação:** slide 15.

Remove Paradise do percurso fazendo Yellow apontar diretamente para Believer, sem deslocar os demais nós.


```python
# Exemplo 06 - Removendo Paradise sem quebrar a playlist
# Conceito: remover um nó reconectando o elemento anterior ao sucessor.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


def exibir(inicio):
    atual = inicio

    while atual is not None:
        print(atual.musica, end="")
        if atual.proximo is not None:
            print(" -> ", end="")
        atual = atual.proximo

    print(" -> None")


imagine = No("Imagine")
yellow = No("Yellow")
paradise = No("Paradise")
believer = No("Believer")

imagine.proximo = yellow
yellow.proximo = paradise
paradise.proximo = believer

print("=== ANTES DA REMOÇÃO ===")
exibir(imagine)

yellow.proximo = paradise.proximo

print("\n=== DEPOIS DA REMOÇÃO ===")
exibir(imagine)

print("\nParadise não faz mais parte do percurso iniciado em Imagine.")
```

## Exemplo 07 — Playlist encadeada em ação

**Origem na apresentação:** slide 16.

Integra criação de nós, referência de início, inserção, remoção e percurso em uma classe que representa a playlist completa.


```python
# Exemplo 07 - Playlist encadeada em ação
# Conceitos: classe No, início da lista, inserção, remoção e percurso.

class No:
    def __init__(self, musica):
        self.musica = musica
        self.proximo = None


class PlaylistEncadeada:
    def __init__(self):
        self.inicio = None

    def adicionar_fim(self, musica):
        novo = No(musica)

        if self.inicio is None:
            self.inicio = novo
            return

        atual = self.inicio

        while atual.proximo is not None:
            atual = atual.proximo

        atual.proximo = novo

    def inserir_apos(self, musica_referencia, nova_musica):
        atual = self.inicio

        while atual is not None:
            if atual.musica == musica_referencia:
                novo = No(nova_musica)
                novo.proximo = atual.proximo
                atual.proximo = novo
                return True

            atual = atual.proximo

        return False

    def remover(self, musica):
        atual = self.inicio
        anterior = None

        while atual is not None:
            if atual.musica == musica:
                if anterior is None:
                    self.inicio = atual.proximo
                else:
                    anterior.proximo = atual.proximo

                return True

            anterior = atual
            atual = atual.proximo

        return False

    def exibir(self):
        atual = self.inicio

        while atual is not None:
            print(atual.musica, end="")

            if atual.proximo is not None:
                print(" -> ", end="")

            atual = atual.proximo

        print(" -> FIM")


playlist = PlaylistEncadeada()

playlist.adicionar_fim("Imagine")
playlist.adicionar_fim("Yellow")
playlist.adicionar_fim("Believer")

print("=== PLAYLIST INICIAL ===")
playlist.exibir()

playlist.inserir_apos("Yellow", "Paradise")

print("\n=== APÓS A INSERÇÃO DE PARADISE ===")
playlist.exibir()

playlist.remover("Yellow")

print("\n=== APÓS A REMOÇÃO DE YELLOW ===")
playlist.exibir()
```

## Síntese

Os exemplos mostram que, em uma lista simplesmente encadeada:

- cada elemento é representado por um nó;
- `proximo` mantém a referência para o sucessor;
- `inicio` permite acessar o primeiro nó;
- o percurso segue as referências até `None`;
- inserir significa ajustar referências para incluir um novo nó;
- remover significa reconectar os nós para retirar um elemento do caminho.
